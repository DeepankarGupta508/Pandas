## 10.Concatenate and Merge Dataframe (JOINS)

```python
row1 = pd.DataFrame({"ID": [1, 2, 3], "Name": ["A", "B", "C"]})
row1
```

```python
row2 = pd.DataFrame({"ID": [1, 2, 2, 3], "Score": ["88", "96", "77", "79"]})
row2
```

```python
pd.concat([row1, row2])     # vertical / row level / top on top
```

```python
pd.concat([row1, row2], axis=1)     # Horizontal / column level / side by side
```

```python
# Merge = Join
pd.merge(row1, row2, how="inner", on= "ID")
```

```python
pd.merge(row1, row2, how="inner", left_on="ID", right_on="ID")
```
