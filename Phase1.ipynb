## 6.Rows and columns - Operations (Add, Update, Delete)

```python
load_data
```

```python
#add new column
load_data["Team"] = ["CEO", "HR", "Marketing Head", "Data Analys"]
load_data
```

```python
load_data["Bonus"] = load_data["Monthly_salary"] * 0.2
load_data
```

```python
# add new row
load_data.loc[len(load_data)] = ["Deepankar", 25, 150000, "chairmain", 30000]
load_data
```

```python
len(load_data)
```

```python
# update value  -   using index name
load_data.loc[0, "Monthly_salary"] = 50000 
load_data
```

```python
# update value  -   using column value
load_data.loc[load_data.Name == "Deepankar", "Monthly_salary"] = 210000
load_data
load_data.loc[load_data.Name == "Deepankar", "Team"] = "ML Engineer"
load_data
```

```python
# Delete value - row and column values
load_data.drop(load_data[load_data.Name == "Mayank"].index)     #Drop row, axis=0
```

```python
load_data.drop(1, axis=0)   # index name
```

```python
load_data.drop("Bonus", axis=1, inplace=True)
```
